```python
# List : Use When data can change (mutable).

fruits = ["Apple","Banana","Mango"]
fruits.append("Orange")
print(fruits)

# Tuple : Use when data should not change (immutable).

colors = ("Red","Green","Blue")
print(colors)
```

```python
numbers = [10,20,30,40,50]

print(numbers[0])
print(numbers[2])
print(numbers[-1])
```

```python
lst = [1,2,3,4]

print(tuple(lst))
print(set(lst))
print(str(lst))
```

```python
numbers = [10,20,30,40]
print(len(numbers))
```

```python
a = [1,2]
a.append([3,4])
print(a)

b = [1,2]
b.extend([3,4])
print(b)
```
